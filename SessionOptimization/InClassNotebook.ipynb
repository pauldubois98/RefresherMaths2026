We are studying the function $f(x) = x^2 -x + 3$.

```python
def f(x):
    return x**2 - x + 3

f(1)
```

```python
xs = np.linspace(-5, 5, 101)
ys = f(xs)

plt.plot(xs, ys, '-')
```

## Grid Search

```python
x_min = -1.75
y_min = f(x_min)
for k in range(0, 8):
    x = -1.75 + k * 0.5
    y = f(x)
    if y < y_min:
        x_min = x
        y_min = y
print(x_min, y_min)
```

```python
a = -5
b = 5
for i in range(6):
    print(f'[{a};{b}] f({a})= {f(a)} f({b})= {f(b)}')
    if f(a) > f(b):
        a = (a+b)/2
    else:
        b = (a+b)/2
```

## Dichotomy

$x \in [a,b]$, $y \in [c,d]$

```python
a = -5 ; b = 5
c = -5 ; d = 5
for i in range(7):
    print(f'x range: [{a:f},{b:f}] ; y range: [{c:f},{d:f}] ;\
          f({(a+b/2)}, {(c+d)/2}) = {f( (a+b)/2, (c+d)/2 )}')
    if f(a,c) + f(a,d) > f(b,c) + f(b,d):
        a_ = (a+b) / 2
        b_ = b
    else:
        a_ = a
        b_ = (a+b) / 2
    if f(a,c) + f(b,c) > f(a,d) + f(b,d):
        c_ = (c+d) / 2
        d_ = d
    else:
        c_ = c
        d_ = (c+d) / 2
    a = a_ ; b = b_ ; c = c_ ; d = d_
```

## Gradient Descent

$f(x,y) = (x+y-1)^2 + \frac{1}{5} (x-y)^2$

So $\frac{\partial f(x,y)}{\partial x} = 2(x+y-1) + \frac{2}{5} (x-y)$

and $\frac{\partial f(x,y)}{\partial y} = 2(x+y-1) - \frac{2}{5} (x-y)$

```python
def dfx(x,y):
    return 2*(x+y-1) + 0.2*(x-y)

def dfy(x,y):
    return 2*(x+y-1) - 0.2*(x-y)
```

```python
x = 2.5
y = 3.5
lamb = 0.3
for i in range(7):
    print(f'(x_{i},y_{i}) = ({x:f},{y:f}) ; f(x,y)={f(x,y)}')
    x_ = x - lamb * dfx(x,y)
    y_ = y - lamb * dfy(x,y)
    x = x_ ; y = y_
```

```python
x = 2.5
y = 3.5
lamb = 0.1
for i in range(7):
    print(f'(x_{i},y_{i}) = ({x:f},{y:f}) ; f(x,y)={f(x,y)}')
    x_ = x - lamb * dfx(x,y)
    y_ = y - lamb * dfy(x,y)
    x = x_ ; y = y_
```
